#### La columna N° estaba enumerada mediante numeración automática en word, por tanto no se imprimió como texto, debido a que el texto es inexistente

```python
alumnos = []
for fila in tabla.rows[1:]:
    nombre = fila.cells[1].text
    alumnos.append(nombre)
```

```python
alumnos
```

###### Se utiliza el estandar: 'APELLIDO, Nombre'

```python
alumnos_sospechosos = []

for alumno in alumnos:
    if any(caracter.isdigit() for caracter in alumno) or "°" in alumno or "/" in alumno or alumno != alumno.strip():
        alumnos_sospechosos.append(alumno)

alumnos_sospechosos
```

```python
cont = 0 

for alumno in alumnos:
    if alumno != alumno.strip():
        cont += 1
cont
```

```python
for caracter in "MARTINEZ, Sofia Isabella a 1º2º 03/04/25": 
    print(caracter,ord(caracter)) 
```

#### Limpieza del contenido de alumnos en una nueva variable
Sólo se quitaron espacios, se conserva el dato sucio 'MARTINEZ, Sofia Isabella a 1º2º 03/04/25', como base de observación

```python
alumnos_limpios = []

for alumno in alumnos:
    alumno_limpio = alumno.strip()
    alumnos_limpios.append(alumno_limpio)
alumnos_limpios
```

```python
for original, limpio in zip(alumnos, alumnos_limpios): 
    if original != limpio:
        print("ORIGINAL:", original)
        print("LIMPIO:", limpio)
```

##### Creación de listas y desfragmentación del nombre del alumno
Se separa "APELLIDO, Nombre" de " metadatos" y se guardan ambos en dos listas por separado, una lista para incorporarse a la nueva sección de alumnos
corregida, y la otra como observación para futuros casos.

```python
alumnos_corregidos = []
observaciones_marzo = []

for alumno in alumnos_limpios:
    if " a " in alumno and any(caracter.isdigit() for caracter in alumno):
        partes = alumno.split(" a", 1)
        nombre_limpio = partes[0]
        observacion = "a " + partes[1]
    else: 
        nombre_limpio = alumno
        observacion = None
    alumnos_corregidos.append(nombre_limpio)
    observaciones_marzo.append(observacion)
```

```python
for nombre, obs in zip(alumnos_corregidos,observaciones_marzo): ## Comprobación de funcionamiento de la desfragmentación del nombre
    if obs is not None:
        print("NOMBRE:", nombre)
        print("OBS:", obs)
```

```python
alumnos_corregidos
```

#### Creación del dataframe para la fuente Marzo

```python
df_marzo = pd.DataFrame({
    "nombre_completo":alumnos_corregidos,
    "observacion_marzo":observaciones_marzo,
    "curso":"1",
    "division":"1",
    "anio":"2025",
    "fuente":"Marzo",
    "archivo_origen":"1°1° 2025 Marzo.docx"
})
```

### 1.4 - Cruce de dataframes del curso 1°1° para validación de datos

#### 1.4.2 - df_datos vs df_legajos

**Objetivo del cruce:**
- ¿Tienen la misma cantidad de alumnos?
